```python
import pandas as pd

inventory = pd.read_csv(
    "../datasets/inventory/Inventory_Transactions.csv"
)

stock = pd.read_csv(
    "../datasets/stock/synthetic_stock_data.csv"
)

medicine_price = pd.read_csv(
    "../datasets/products/medicine_price_dataset.csv"
)

products = pd.read_csv(
    "../datasets/products/pharma_products_dataset.csv"
)
```

```python
def profile_dataset(df, name):

    print("="*60)

    print(f"Dataset: {name}")

    print("="*60)

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nMissing Values:")
    print(df.isnull().sum().sort_values(
        ascending=False
    ).head(10))

    print("\nDuplicates:")
    print(df.duplicated().sum())

    print("\nData Types:")
    print(df.dtypes)

    print("\n")
```

```python
profile_dataset(
    sales,
    "Drug Sales"
)

profile_dataset(
    inventory,
    "Inventory"
)

profile_dataset(
    stock,
    "Stock"
)

profile_dataset(
    medicine_price,
    "Medicine Price"
)

profile_dataset(
    products,
    "Products"
)
```
